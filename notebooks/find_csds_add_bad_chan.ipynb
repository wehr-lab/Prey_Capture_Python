```python
import os
import shutil

def copy_lfp_figs_to_csd_plots(base_dir):
    """
    Copies all '*-LFP-figs.pdf' files found one level deep under base_dir
    to base_dir/CSD_Plots. Returns a set of directories that contained matching files.
    """
    suffix = '-LFP-figs.pdf'
    csd_plots_dir = os.path.join(base_dir, 'CSD_Plots')
    os.makedirs(csd_plots_dir, exist_ok=True)

    contributing_dirs = set()

    for entry in os.scandir(base_dir):
        if entry.is_dir() and entry.name != 'CSD_Plots':
            sub_dir = entry.path
            for filename in os.listdir(sub_dir):
                if filename.endswith(suffix):
                    contributing_dirs.add(sub_dir)
                    src_path = os.path.join(sub_dir, filename)
                    dest_path = os.path.join(csd_plots_dir, filename)

                    # Skip if already copied
                    if os.path.exists(dest_path) and os.path.getsize(src_path) == os.path.getsize(dest_path):
                        print(f"Skipped (already copied): {filename}")
                        continue

                    # Rename if conflict with different content
                    base, ext = os.path.splitext(filename)
                    counter = 1
                    while os.path.exists(dest_path):
                        if os.path.getsize(src_path) == os.path.getsize(dest_path):
                            print(f"Skipped (duplicate under different name): {os.path.basename(dest_path)}")
                            break
                        dest_path = os.path.join(csd_plots_dir, f"{base}_{counter}{ext}")
                        counter += 1
                    else:
                        shutil.copy2(src_path, dest_path)
                        print(f"Copied: {src_path} -> {dest_path}")

    return contributing_dirs
```

```python
base_dir = '/Volumes/Projects/PreyCapture/Rig3Phys'
session_dirs_with_plots = copy_lfp_figs_to_csd_plots(base_dir)
```

```python
def create_bad_channel_files(session_dirs):
    """
    For each session directory, prompt user to enter bad channels,
    then save them to 'bad_channels.txt' in that directory.
    """
    for session_dir in sorted(session_dirs):
        session_name = os.path.basename(session_dir)
        label_path = os.path.join(session_dir, 'bad_channels.txt')

        # Skip if it already exists
        if os.path.exists(label_path):
            print(f"Skipping {session_name} (already has bad_channels.txt)")
            continue

        # Prompt for input
        user_input = input(f"Enter bad channels for {session_name} (comma-separated, or leave blank to skip): ").strip()

        if not user_input:
            print(f"Skipped {session_name} (no channels entered)")
            continue

        # Parse and clean input
        channels = [ch.strip() for ch in user_input.split(',') if ch.strip().isdigit()]

        # Write to file
        with open(label_path, 'w') as f:
            f.write("# Manually entered bad channels for this session:\n")
            for ch in channels:
                f.write(f"{ch}\n")

        print(f"Created: {label_path}")
```

```python
create_bad_channel_files(session_dirs_with_plots)
```
